```python
import pandas as pd

df = pd.read_csv('data/raw/customers_orgs.csv')
df.head()
```

```python
#Convertir signup_date a formato fecha, venía como obj
df['signup_date'] = pd.to_datetime(df['signup_date'])
df['signup_date']
```

```python
df.isnull().sum()
```

```python
df['nps_score'].describe()
```

```python
df[df['nps_score'] > 100]
```

Se detectó un valor de 101 en `nps_score`, matemáticamente imposible ya que el NPS
tiene un rango válido de -100 a 100. Se interpretó como error de carga y se convirtió
a NaN, sumándose al resto de los valores faltantes de la columna.

```python
df.loc[df['nps_score'] > 100, 'nps_score'] = None
```

```python
df[df['nps_score'] < 0]
```

```python
pd.crosstab(
    df["nps_score"].isnull(),
    df["lifecycle_stage"]
)
```

```python
print("Sin NPS:")
print(df[df['nps_score'].isnull()]['signup_date'].describe())

print("\nCon NPS:")
print(df[df['nps_score'].notnull()]['signup_date'].describe())
```

Descarto que los valores nulos de NPS corresponden a clientes ingresados recientemente.Dejo los valores nulos para evaluar luego
