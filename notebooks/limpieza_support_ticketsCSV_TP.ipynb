```python
from google.colab import drive
import pandas as pd
import numpy as np
import datetime

df = pd.read_csv('support_tickets.csv')

print("=== Estructura y tipos de datos iniciales ===")
print(df.info())
```

```python
print("\n=== Cantidad de valores nulos ===")
print(df.isnull().sum())
```

```python


# columnas de fecha a formato Datetime
df['created_at'] = pd.to_datetime(df['created_at'])
df['resolved_at'] = pd.to_datetime(df['resolved_at'])

# eliminar espacios en blanco y convertir a minúsculas
df['category'] = df['category'].str.lower().str.strip()
df['severity'] = df['severity'].str.lower().str.strip()

# calificaciones de CSAT inválidas fuera del rango de 1.0 a 5.0
# Valores como 0.0, 6.0 o 7.0 se reemplazan por NaN
df.loc[(df['csat'] < 1.0) | (df['csat'] > 5.0), 'csat'] = np.nan

# inconsistencia lógica Tickets abiertos no pueden tener CSAT
# Si 'resolved_at' es nulo (ticket abierto), el CSAT debe ser NaN (nulo)
df.loc[df['resolved_at'].isnull(), 'csat'] = np.nan

# datos limpios en un nuevo archivo CSV
df.to_csv('support_tickets_clean.csv', index=False)

print(df.info())
```
