```python
import pandas as pd
#Se carga el archivo
df = pd.read_csv('data/raw/users.csv')
df.head(20)
```

```python
#Verifico el tipo de datos
print(df.dtypes)
```

```python
#Transformo las columnas de fecha a datatime
date_cols = ['created_at', 'last_login']
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
#Verifico duplicado
total_duplicados = df.duplicated().sum()
print(f"Filas duplicadas totales: {total_duplicados}")

#Mostrar las filas duplicadas para inspeccionarlas (keep=False muestra todas las apariciones)
filas_duplicadas = df[df.duplicated(keep=False)]
print(filas_duplicadas)
```

```python
#Verifico si existen duplicados de usuarios o de mail
user_duplicados = df['user_id'].duplicated().sum()
email_duplicados = df['email'].duplicated().sum()

print(f"IDs duplicados: {user_duplicados}")
print(f"Emails duplicados: {email_duplicados}")

#Filas que comparten un mismo correo
df[df['email'].duplicated(keep=False)]
```

```python
#Se verifican y corrigen incoherencias como ejemplo la fila N°5 donde el ultimo ingreso es antes que la creación de la cuenta

mask_invalid_date = df['last_login'] < df['created_at']
df.loc[mask_invalid_date, 'last_login'] = pd.NaT
```

```python
#Normalización de texto
df['email'] = df['email'].astype(str).str.strip().str.lower()
df['role'] = df['role'].astype(str).str.strip()
```

```python
#Guardo la versión limpia

df.to_csv('data/clean/users_limpio.csv', index=False)
print("--- Limpieza completada con éxito ---")
print(f"Total registros: {len(df)}")
print(f"Fechas incoherentes corregidas a nulo: {mask_invalid_date.sum()}")
print(f"Registros sin último inicio de sesión (NaT): {df['last_login'].isna().sum()}")
```
